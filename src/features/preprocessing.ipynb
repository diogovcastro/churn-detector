### Remoção de registros com TotalCharges ausente

Conforme identificado em `validation.ipynb`, 11 registros (0,16% da base)
possuem `TotalCharges` ausente. Optou-se por removê-los devido à pequena
proporção, mantendo 7.032 registros em `df_clean`.

Todos os registros removidos têm `tenure = 0`, portanto a exclusão afeta
especificamente esse grupo de clientes.

```python
import os
from pathlib import Path

import pandas as pd
from dotenv import find_dotenv, load_dotenv
```

```python
arquivo_env = find_dotenv(usecwd=True)

if not arquivo_env:
    raise FileNotFoundError("Arquivo .env não encontrado.")

load_dotenv(arquivo_env)
projeto = Path(arquivo_env).parent

# Carrega a base original.
caminho_raw = projeto / os.environ["RAW_CSV_PATH"]
df_raw = pd.read_csv(caminho_raw, dtype={"customerID": "string"})

if df_raw.empty:
    raise ValueError("A base raw está vazia.")

# Identifica nulos, campos vazios e campos contendo apenas espaços.
total_charges = df_raw["TotalCharges"].astype("string").str.strip()
ausentes = total_charges.isna() | total_charges.eq("").fillna(False)

quantidade_removida = int(ausentes.sum())
percentual_removido = ausentes.mean() * 100

# Mantém somente os registros com TotalCharges preenchido.
df_clean = df_raw.loc[~ausentes].copy().reset_index(drop=True)

print(f"Registros originais: {len(df_raw)}")
print(
    f"Registros removidos: {quantidade_removida} "
    f"({percentual_removido:.2f}%)"
)
print(f"Registros restantes: {len(df_clean)}")
```
